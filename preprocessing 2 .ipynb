````python
#cell1
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

import os
import pickle
import numpy as np
import pandas as pd
from pathlib import Path

BASE          = "/content/drive/MyDrive/RAIM_Project_v2/run_05_age_med_years_med"
synthea_path  = f"{BASE}/raw/csv"
output_path   = f"{BASE}/processed"
matrices_dir  = f"{BASE}/matrices"

CHUNK_SIZE = 50_000

os.makedirs(output_path, exist_ok=True)
os.makedirs(matrices_dir, exist_ok=True)

print("Paths ready.")
print("matrices_dir:", matrices_dir)
````

````python
# cell 2
ENC_COLS = ['Id', 'PATIENT', 'ENCOUNTERCLASS', 'START', 'STOP']
PAT_COLS = ['Id', 'GENDER', 'RACE', 'ETHNICITY', 'BIRTHDATE']

enc_chunks = []
for chunk in pd.read_csv(f"{synthea_path}/encounters.csv", usecols=ENC_COLS, chunksize=CHUNK_SIZE):
    enc_chunks.append(chunk[chunk['ENCOUNTERCLASS'] == 'inpatient'])
encounters_df = pd.concat(enc_chunks, ignore_index=True).rename(columns={
    'Id': 'hadm_id', 'PATIENT': 'subject_id',
    'START': 'admittime', 'STOP': 'dischtime',
    'ENCOUNTERCLASS': 'admission_type',
})
del enc_chunks

inpatient_subject_ids = set(encounters_df['subject_id'])

pat_chunks = []
for chunk in pd.read_csv(f"{synthea_path}/patients.csv", usecols=PAT_COLS, chunksize=CHUNK_SIZE):
    pat_chunks.append(chunk[chunk['Id'].isin(inpatient_subject_ids)])
patients_df = pd.concat(pat_chunks, ignore_index=True).rename(columns={
    'Id': 'subject_id', 'GENDER': 'gender',
    'RACE': 'race', 'ETHNICITY': 'ethnicity', 'BIRTHDATE': 'birthdate',
})
del pat_chunks

df = encounters_df.merge(patients_df, on='subject_id', how='left')
del encounters_df, patients_df

for col in ('admittime', 'dischtime', 'birthdate'):
    df[col] = pd.to_datetime(df[col], errors='coerce').dt.tz_localize(None)

df = df.dropna(subset=['admittime', 'dischtime']).copy()
df['LOS'] = (df['dischtime'] - df['admittime']).dt.days
df['anchor_age'] = (df['admittime'] - df['birthdate']).dt.days // 365
df = df[(df['LOS'] > 0) & (df['LOS'] <= 500)].copy()

# 1990 filter
df = df[df['admittime'].dt.year >= 1990].copy()

# LLOS cohort filter
los_mean = df['LOS'].mean()
los_std  = df['LOS'].std(ddof=0)
los_threshold = los_mean + 2 * los_std
print(f"LOS mean: {los_mean:.2f}  std: {los_std:.2f}  threshold: {los_threshold:.2f}")

llos_df = df[df['LOS'] > los_threshold].copy()
cohort_subj_ids = set(llos_df['subject_id'])
df = df[df['subject_id'].isin(cohort_subj_ids)].copy()
df = df.rename(columns={'LOS': 'target'})
cohort_hadm_ids = set(df['hadm_id'])

print(f"Encounters after 1990 filter + LLOS cohort: {len(df)}")
print(f"Unique patients: {df['subject_id'].nunique()}")
print(df['target'].describe())
````

````python
# cell 3
OBS_COLS = ['PATIENT', 'ENCOUNTER', 'DATE', 'DESCRIPTION', 'VALUE', 'UNITS']

obs_chunks = []
for chunk in pd.read_csv(f"{synthea_path}/observations.csv", usecols=OBS_COLS, chunksize=CHUNK_SIZE):
    obs_chunks.append(chunk[chunk['ENCOUNTER'].isin(cohort_hadm_ids)])
observations_df = pd.concat(obs_chunks, ignore_index=True).rename(columns={
    'PATIENT': 'subject_id', 'ENCOUNTER': 'hadm_id',
    'DATE': 'obs_date', 'DESCRIPTION': 'obs_description',
    'VALUE': 'obs_value', 'UNITS': 'obs_units',
})
del obs_chunks

observations_df['obs_date'] = pd.to_datetime(
    observations_df['obs_date'], errors='coerce'
).dt.tz_localize(None)

print(f"Observations loaded: {len(observations_df)}")
````

````python
# cell 4 - vital snapshot into static (replaces 12x5 matrix)
VITAL_CHANNELS = {
    'Heart rate':               'hr',
    'Systolic Blood Pressure':  'sbp',
    'Diastolic Blood Pressure': 'dbp',
    'Respiratory rate':         'rr',
    'Body Weight':              'weight',
}
CHANNELS = ['hr', 'sbp', 'dbp', 'rr', 'weight']

obs_vitals = observations_df[
    observations_df['obs_description'].isin(VITAL_CHANNELS)
].copy()
obs_vitals['channel'] = obs_vitals['obs_description'].map(VITAL_CHANNELS)
obs_vitals['obs_value'] = pd.to_numeric(obs_vitals['obs_value'], errors='coerce')

# Take mean per encounter per channel (handles rare cases with multiple readings)
vital_snapshot = (
    obs_vitals.groupby(['hadm_id', 'channel'])['obs_value']
    .mean().unstack(fill_value=np.nan)
)
vital_snapshot.columns = [f'vital_{c}' for c in vital_snapshot.columns]
vital_snapshot = vital_snapshot.reindex(columns=[f'vital_{c}' for c in CHANNELS])

print(f"Vital snapshots available: {len(vital_snapshot)} / {len(df)} encounters")
print(f"Coverage per channel:\n{vital_snapshot.notna().mean()}")
````

````python
# cell 5 - absolute 12-hour window guidance matrix
LAB_SKIP = {
    'Heart rate', 'Systolic Blood Pressure', 'Diastolic Blood Pressure',
    'Respiratory rate', 'Body Weight', 'Body Height',
    'Body mass index (BMI) [Ratio]',
    'Pain severity - 0-10 verbal numeric rating [Score] - Reported',
    'DALY', 'QOLS', 'QALY'
}

# Compute hours from admit for all observations
obs_with_admit = observations_df.merge(
    df[['hadm_id', 'admittime', 'dischtime']], on='hadm_id', how='inner'
)
obs_with_admit['hours_from_admit'] = (
    (obs_with_admit['obs_date'] - obs_with_admit['admittime'])
    .dt.total_seconds() / 3600
)
obs_with_admit = obs_with_admit[
    (obs_with_admit['hours_from_admit'] >= 0) &
    (obs_with_admit['obs_date'] <= obs_with_admit['dischtime'])
].copy()

# Absolute 12-hour bin index
obs_with_admit['hour_bin'] = (obs_with_admit['hours_from_admit'] / 12).astype(int)

# Lab flags per absolute window
labs_binned = obs_with_admit[
    ~obs_with_admit['obs_description'].isin(LAB_SKIP)
].copy()
lab_flags = (
    labs_binned.groupby(['hadm_id', 'hour_bin']).size()
    .reset_index(name='count')
)

# Med flags per absolute window
med_chunks = []
for chunk in pd.read_csv(f"{synthea_path}/medications.csv",
                          usecols=['START', 'ENCOUNTER'], chunksize=CHUNK_SIZE):
    chunk = chunk[chunk['ENCOUNTER'].isin(cohort_hadm_ids)].copy()
    med_chunks.append(chunk)
meds_df = pd.concat(med_chunks, ignore_index=True)
meds_df.columns = ['med_start', 'hadm_id']
meds_df['med_start'] = pd.to_datetime(meds_df['med_start'], errors='coerce').dt.tz_localize(None)
del med_chunks

meds_in_enc = meds_df.merge(df[['hadm_id', 'admittime', 'dischtime']], on='hadm_id', how='inner')
meds_in_enc = meds_in_enc[
    (meds_in_enc['med_start'] >= meds_in_enc['admittime']) &
    (meds_in_enc['med_start'] <= meds_in_enc['dischtime'])
].copy()
meds_in_enc['hours_from_admit'] = (
    (meds_in_enc['med_start'] - meds_in_enc['admittime']).dt.total_seconds() / 3600
)
meds_in_enc['hour_bin'] = (meds_in_enc['hours_from_admit'] / 12).astype(int)

med_flags = (
    meds_in_enc.groupby(['hadm_id', 'hour_bin']).size()
    .reset_index(name='count')
)

print(f"Lab events binned: {len(labs_binned)}")
print(f"Med events binned: {len(meds_in_enc)}")
````

````python
# cell 6 - patient-level condition history (ICD-10)
COND_COLS = ['START', 'STOP', 'PATIENT', 'ENCOUNTER', 'CODE']

cond_chunks = []
for chunk in pd.read_csv(f"{synthea_path}/conditions_icd10.csv",
                          usecols=COND_COLS, chunksize=CHUNK_SIZE):
    chunk = chunk[chunk['PATIENT'].isin(cohort_subj_ids)].copy()
    chunk['START'] = pd.to_datetime(chunk['START'], errors='coerce').dt.tz_localize(None)
    chunk['is_ongoing'] = chunk['STOP'].isna()
    cond_chunks.append(chunk)

conditions_raw = pd.concat(cond_chunks, ignore_index=True).rename(columns={
    'PATIENT': 'subject_id',
    'CODE':    'icd_code',
    'START':   'cond_start',
})
del cond_chunks

# Join to encounters to get hadm_id and admittime
cond_with_enc = conditions_raw.merge(
    df[['hadm_id', 'subject_id', 'admittime']], on='subject_id', how='inner'
)
del conditions_raw

# Only conditions that existed before or at admission (no leakage)
cond_at_admit = cond_with_enc[
    cond_with_enc['cond_start'] <= cond_with_enc['admittime']
].copy()
del cond_with_enc

print(f"Condition rows after patient-level join + pre-admission filter: {len(cond_at_admit)}")
print(f"Encounters covered: {cond_at_admit['hadm_id'].nunique()} / {len(df)}")
print(f"Unique ICD codes: {cond_at_admit['icd_code'].nunique()}")
print(f"Avg conditions per encounter: {len(cond_at_admit) / cond_at_admit['hadm_id'].nunique():.1f}")
print(f"\nTop 15 ICD codes:\n{cond_at_admit['icd_code'].value_counts().head(15)}")

# One-hot encode
cond_at_admit['icd_code'] = 'icd_' + cond_at_admit['icd_code'].astype(str)
icd_dummies = (
    cond_at_admit.groupby(['hadm_id', 'icd_code'])
    .size().unstack(fill_value=0).clip(upper=1).astype('int8')
    .reset_index()
)
print(f"\nICD dummies: {icd_dummies.shape[0]} encounters x {icd_dummies.shape[1]-1} codes")

# Static feature df: demographics + ICD
static_df = df[['hadm_id', 'subject_id', 'admittime', 'target',
                 'anchor_age', 'gender', 'race', 'ethnicity']].copy()
static_df = static_df.merge(icd_dummies, on='hadm_id', how='left')
del icd_dummies

icd_cols = [c for c in static_df.columns if c.startswith('icd_')]
for col in icd_cols:
    static_df[col] = static_df[col].fillna(0).astype('int8')

print(f"\nStatic feature df shape: {static_df.shape}")
print(f"ICD columns: {len(icd_cols)}")
````

````python
# cell 6b - add vital snapshot to static_df
vital_cols = [f'vital_{c}' for c in CHANNELS]
static_df = static_df.merge(
    vital_snapshot.reset_index(), on='hadm_id', how='left'
)
static_df['has_vitals'] = static_df[vital_cols].notna().any(axis=1).astype('int8')
for col in vital_cols:
    static_df[col] = static_df[col].fillna(0).astype('float32')

print(f"Static feature df shape after vitals: {static_df.shape}")
# Should now be (2578, 50)
````

````python
# cell 6c - patient-level procedure history (ICD-10-PCS)
PROC_COLS = ['START', 'PATIENT', 'ENCOUNTER', 'CODE']

proc_chunks = []
for chunk in pd.read_csv(f"{synthea_path}/procedures_icd10.csv",
                          usecols=PROC_COLS, chunksize=CHUNK_SIZE):
    chunk = chunk[chunk['PATIENT'].isin(cohort_subj_ids)].copy()
    chunk['START'] = pd.to_datetime(chunk['START'], errors='coerce').dt.tz_localize(None)
    proc_chunks.append(chunk)

procedures_raw = pd.concat(proc_chunks, ignore_index=True).rename(columns={
    'PATIENT': 'subject_id',
    'CODE':    'proc_code',
    'START':   'proc_start',
})
del proc_chunks

# Join to encounters to get hadm_id and admittime
proc_with_enc = procedures_raw.merge(
    df[['hadm_id', 'subject_id', 'admittime']], on='subject_id', how='inner'
)
del procedures_raw

# Only procedures before admission (historical) — no leakage
proc_at_admit = proc_with_enc[
    proc_with_enc['proc_start'] <= proc_with_enc['admittime']
].copy()
del proc_with_enc

print(f"Procedure rows after patient-level join + pre-admission filter: {len(proc_at_admit)}")
print(f"Encounters covered: {proc_at_admit['hadm_id'].nunique()} / {len(df)}")
print(f"Unique procedure codes: {proc_at_admit['proc_code'].nunique()}")
print(f"Avg procedures per encounter: {len(proc_at_admit) / max(proc_at_admit['hadm_id'].nunique(), 1):.1f}")
print(f"\nTop 15 procedure codes:\n{proc_at_admit['proc_code'].value_counts().head(15)}")

# One-hot encode
proc_at_admit['proc_code'] = 'proc_' + proc_at_admit['proc_code'].astype(str)
proc_dummies = (
    proc_at_admit.groupby(['hadm_id', 'proc_code'])
    .size().unstack(fill_value=0).clip(upper=1).astype('int8')
    .reset_index()
)
print(f"\nProc dummies: {proc_dummies.shape[0]} encounters x {proc_dummies.shape[1]-1} codes")

# Merge into static_df
static_df = static_df.merge(proc_dummies, on='hadm_id', how='left')
del proc_dummies

proc_cols = [c for c in static_df.columns if c.startswith('proc_')]
for col in proc_cols:
    static_df[col] = static_df[col].fillna(0).astype('int8')

print(f"\nStatic feature df shape after procedures: {static_df.shape}")
print(f"Procedure columns: {len(proc_cols)}")
````

````python
# cell 7 - per-window sequences

WINDOW_HOURS = 12

# Index guidance flags for fast lookup
lab_flags_lookup = lab_flags.groupby('hadm_id')['hour_bin'].apply(set).to_dict()
med_flags_lookup = med_flags.groupby('hadm_id')['hour_bin'].apply(set).to_dict()

# Static feature vector per encounter (drop non-numeric identifier cols)
static_feat_cols = [c for c in static_df.columns if c not in
                    ['hadm_id', 'subject_id', 'admittime', 'target', 'gender', 'race', 'ethnicity']]

# Encode categoricals
static_df['gender_enc']    = (static_df['gender'] == 'M').astype('int8')
static_df['race_enc']      = pd.Categorical(static_df['race']).codes.astype('int16')
static_df['ethnicity_enc'] = pd.Categorical(static_df['ethnicity']).codes.astype('int16')

static_feat_cols = (
    ['anchor_age', 'gender_enc', 'race_enc', 'ethnicity_enc', 'has_vitals']
    + [f'vital_{c}' for c in CHANNELS]
    + [c for c in static_df.columns if c.startswith('icd_')]
)

static_lookup = static_df.set_index('hadm_id')[static_feat_cols]

sequences = {}   # hadm_id -> {'X': (T, 52), 'Y': (T,), 'los': float}
manifest_rows = []

for _, enc in df[['hadm_id', 'subject_id', 'admittime', 'target']].iterrows():
    hid      = enc['hadm_id']
    los_days = float(enc['target'])
    los_hrs  = los_days * 24
    n_windows = max(1, int(los_hrs / WINDOW_HOURS))

    static_vec = static_lookup.loc[hid].values.astype('float32')  # (50,)
    lab_bins   = lab_flags_lookup.get(hid, set())
    med_bins   = med_flags_lookup.get(hid, set())

    X_seq = []
    Y_seq = []
    for w in range(n_windows):
        lab_active = 1 if w in lab_bins else 0
        med_active = 1 if w in med_bins else 0
        guidance   = np.array([lab_active, med_active], dtype='float32')
        x_w        = np.concatenate([static_vec, guidance])  # (52,)

        remaining_los = los_days - (w * WINDOW_HOURS / 24)
        remaining_los = max(0.0, remaining_los)

        X_seq.append(x_w)
        Y_seq.append(remaining_los)

    sequences[hid] = {
        'X': np.array(X_seq, dtype='float32'),   # (T, 52)
        'Y': np.array(Y_seq, dtype='float32'),   # (T,)
        'los': los_days,
    }
    manifest_rows.append({
        'hadm_id':    hid,
        'subject_id': enc['subject_id'],
        'admittime':  enc['admittime'],
        'los':        los_days,
        'n_windows':  n_windows,
        'admit_year': enc['admittime'].year,
    })

manifest = pd.DataFrame(manifest_rows)

with open(f"{matrices_dir}/sequences.pkl", 'wb') as f:
    pickle.dump(sequences, f)
manifest.to_csv(f"{matrices_dir}/manifest.csv", index=False)

print(f"Saved sequences.pkl  — {len(sequences)} encounters")
print(f"Saved manifest.csv   — {len(manifest)} rows")
print(f"Mean windows per encounter: {manifest['n_windows'].mean():.1f}")
print(f"Max windows: {manifest['n_windows'].max()}")
print(f"Feature dim per window: {list(sequences.values())[0]['X'].shape[1]}")
````

````python
# Cell 8 - Parse note files, extract section closest to each encounter's admittime

import os
import re
import pickle
from datetime import datetime, timedelta

notes_dir = f"{BASE}/raw/notes"

# Build a lookup: subject_id -> note filepath
note_files = os.listdir(notes_dir)
subject_to_notefile = {}
for fname in note_files:
    # filename format: FirstLast_UUID.txt — UUID is everything after last underscore, before .txt
    uuid = fname.replace('.txt', '').split('_')[-1]
    subject_to_notefile[uuid] = os.path.join(notes_dir, fname)

print(f"Note files indexed: {len(subject_to_notefile)}")

# Split a note file into dated sections
DATE_PATTERN = re.compile(r'^\d{4}-\d{2}-\d{2}', re.MULTILINE)

def parse_note_sections(note_text):
    """Returns list of (date, text_chunk) tuples, sorted descending by date (newest first in Synthea)."""
    matches = list(DATE_PATTERN.finditer(note_text))
    sections = []
    for i, match in enumerate(matches):
        date_str = match.group()
        start = match.start()
        end = matches[i+1].start() if i+1 < len(matches) else len(note_text)
        chunk = note_text[start:end].strip()
        try:
            date = datetime.strptime(date_str, '%Y-%m-%d')
            sections.append((date, chunk))
        except:
            continue
    return sections

def find_closest_section(sections, admittime, tolerance_days=3):
    """Find section whose date is closest to admittime, within tolerance."""
    admit_dt = pd.Timestamp(admittime).to_pydatetime().replace(tzinfo=None)
    best = None
    best_delta = timedelta(days=999)
    for (sec_date, sec_text) in sections:
        delta = abs(sec_date - admit_dt)
        if delta < best_delta and delta <= timedelta(days=tolerance_days):
            best_delta = delta
            best = sec_text
    return best

# Build encounter_notes: hadm_id -> note text for that visit
encounter_notes = {}
missing_note = 0
missing_section = 0

for subject_id, group in df.groupby('subject_id'):
    note_path = subject_to_notefile.get(subject_id)
    if note_path is None:
        missing_note += len(group)
        continue

    with open(note_path, 'r', errors='replace') as f:
        note_text = f.read()

    sections = parse_note_sections(note_text)
    if not sections:
        missing_section += len(group)
        continue

    for _, enc in group.iterrows():
        section = find_closest_section(sections, enc['admittime'])
        if section:
            encounter_notes[enc['hadm_id']] = section
        else:
            missing_section += 1

print(f"Encounter notes extracted: {len(encounter_notes)} / {len(df)}")
print(f"Missing note file: {missing_note}")
print(f"No matching section found: {missing_section}")

# Save
with open(f"{matrices_dir}/encounter_notes.pkl", 'wb') as f:
    pickle.dump(encounter_notes, f)
print("Saved encounter_notes.pkl")
````

````python
# Cell 9 - ICD extraction via Groq (Llama 3.1 8B)
import requests
import json
import time
import pickle
import os
import re

GROQ_API_KEY = "YOUR_GROQ_KEY"  # paste your Groq key
API_URL = "https://api.groq.com/openai/v1/chat/completions"
MODEL_ID = "llama-3.1-8b-instant"

HEADERS = {
    "Authorization": f"Bearer {GROQ_API_KEY}",
    "Content-Type": "application/json",
}

SYSTEM_PROMPT = (
    "You are a clinical coding assistant. "
    "Given a clinical note, extract all applicable ICD-10 diagnosis codes. "
    "Return ONLY a valid JSON array of ICD-10 code strings, nothing else. "
    "Example: [\"J18.9\", \"E11.9\", \"I10\"]. "
    "If no codes can be extracted, return []."
)

def extract_icd_codes(note_text, max_chars=3500):
    truncated = note_text[:max_chars]
    payload = {
        "model": MODEL_ID,
        "messages": [
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": truncated}
        ],
        "max_tokens": 200,
        "temperature": 0.0,
    }
    try:
        response = requests.post(API_URL, headers=HEADERS, json=payload, timeout=30)
        response.raise_for_status()
        content = response.json()['choices'][0]['message']['content'].strip()
        content = content.replace('```json', '').replace('```', '').strip()
        codes = json.loads(content)
        if isinstance(codes, list):
            # Deduplicate, strip whitespace, basic format fix
            cleaned = []
            seen = set()
            for c in codes:
                c = str(c).strip()
                # Insert decimal if missing e.g. K014 -> K01.4
                if re.match(r'^[A-Z]\d{3,}$', c):
                    c = c[:3] + '.' + c[3:]
                if c not in seen:
                    seen.add(c)
                    cleaned.append(c)
            return cleaned
        return []
    except Exception as e:
        return []

# Load notes
with open(f"{matrices_dir}/encounter_notes.pkl", 'rb') as f:
    encounter_notes = pickle.load(f)

results_path = f"{matrices_dir}/llm_icd_codes.pkl"
if os.path.exists(results_path):
    with open(results_path, 'rb') as f:
        llm_icd_codes = pickle.load(f)
    print(f"Resuming from {len(llm_icd_codes)} already processed")
else:
    llm_icd_codes = {}

hadm_ids = [hid for hid in encounter_notes if hid not in llm_icd_codes]
print(f"Remaining to process: {len(hadm_ids)}")

SAVE_EVERY = 50

for i, hid in enumerate(hadm_ids):
    codes = extract_icd_codes(encounter_notes[hid])
    llm_icd_codes[hid] = codes

    if (i + 1) % SAVE_EVERY == 0:
        with open(results_path, 'wb') as f:
            pickle.dump(llm_icd_codes, f)
        print(f"[{i+1}/{len(hadm_ids)}] checkpoint saved — last codes: {codes}")

    time.sleep(0.5)  # Groq is fast, lighter rate limit needed

with open(results_path, 'wb') as f:
    pickle.dump(llm_icd_codes, f)

print(f"\nDone. Total encounters with LLM codes: {len(llm_icd_codes)}")
total_codes = sum(len(v) for v in llm_icd_codes.values())
print(f"Total ICD codes extracted: {total_codes}")
print(f"Avg per encounter: {total_codes/max(len(llm_icd_codes),1):.1f}")
````

````python
# Cell 10 - Union LLM ICD codes into static_df, rebuild sequences.pkl

import pickle
import numpy as np

with open(f"{matrices_dir}/llm_icd_codes.pkl", 'rb') as f:
    llm_icd_codes = pickle.load(f)

# Get existing ICD columns in static_df
existing_icd_cols = set(c for c in static_df.columns if c.startswith('icd_'))

# Collect all new codes from LLM not already in static_df
all_llm_codes = set()
for codes in llm_icd_codes.values():
    for c in codes:
        col = 'icd_' + c.replace('.', '_')
        all_llm_codes.add(col)

new_cols = sorted(all_llm_codes - existing_icd_cols)
print(f"Existing ICD columns: {len(existing_icd_cols)}")
print(f"New ICD columns from LLM: {len(new_cols)}")

# Add new columns to static_df, initialized to 0
for col in new_cols:
    static_df[col] = 0

# Set LLM-derived flags per encounter
enriched = 0
for hid, codes in llm_icd_codes.items():
    matches = static_df.index[static_df['hadm_id'] == hid]
    if len(matches) == 0:
        continue
    idx = matches[0]
    for c in codes:
        col = 'icd_' + c.replace('.', '_')
        if col in static_df.columns:
            static_df.at[idx, col] = 1
    enriched += 1

print(f"Encounters enriched: {enriched}")
print(f"static_df shape after enrichment: {static_df.shape}")

# Rebuild static_feat_cols to include new ICD columns
static_feat_cols_enriched = (
    ['anchor_age', 'gender_enc', 'race_enc', 'ethnicity_enc', 'has_vitals']
    + [f'vital_{c}' for c in CHANNELS]
    + [c for c in static_df.columns if c.startswith('icd_')]
    + [c for c in static_df.columns if c.startswith('proc_')]
)

static_lookup_enriched = static_df.set_index('hadm_id')[static_feat_cols_enriched]

# Rebuild sequences
sequences_enriched = {}

for _, enc in df[['hadm_id', 'subject_id', 'admittime', 'target']].iterrows():
    hid      = enc['hadm_id']
    los_days = float(enc['target'])
    los_hrs  = los_days * 24
    n_windows = max(1, int(los_hrs / WINDOW_HOURS))

    static_vec = static_lookup_enriched.loc[hid].values.astype('float32')
    lab_bins   = lab_flags_lookup.get(hid, set())
    med_bins   = med_flags_lookup.get(hid, set())

    X_seq = []
    Y_seq = []
    for w in range(n_windows):
        lab_active = 1 if w in lab_bins else 0
        med_active = 1 if w in med_bins else 0
        guidance   = np.array([lab_active, med_active], dtype='float32')
        x_w        = np.concatenate([static_vec, guidance])

        remaining_los = max(0.0, los_days - (w * WINDOW_HOURS / 24))
        X_seq.append(x_w)
        Y_seq.append(remaining_los)

    sequences_enriched[hid] = {
        'X': np.array(X_seq, dtype='float32'),
        'Y': np.array(Y_seq, dtype='float32'),
        'los': los_days,
    }

# Overwrite sequences.pkl
with open(f"{matrices_dir}/sequences.pkl", 'wb') as f:
    pickle.dump(sequences_enriched, f)

sample = list(sequences_enriched.values())[0]
print(f"\nsequences.pkl overwritten")
print(f"Total encounters: {len(sequences_enriched)}")
print(f"New feature dim per window: {sample['X'].shape[1]}")
````
